```python
#ex:pass/fail
std = {
    "A" : 45,
    "B" : 67,
    "C" : 88,
    "D" : 23}
```

```python
for i,j in std.items():
    if j>50:
        print(i,": Pass")
    else:
        print(i,": Fail")
```
