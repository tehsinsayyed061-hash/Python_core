```python
class BankForm:
    def __init__(self,name,balance):  # phele c1 = BankForm("Tehsin",50000) run
        self.name = name         #name: variable/attributes/instance
        self.balance = balance
        
    def details(self):               # then this c1.details() ////details: methods
        print("Account holder name :",self.name)
        print("Account balance name :",self.balance)
```

```python
c1 = BankForm("Tehsin",50000)
c2 = BankForm("Asfa",4000)
c3 = BankForm("Nive",80000)
```

```python
c1.details() 
c2.details() 
c3.details() 
```

```python
class Circle:
    def __init__(self,radius):  #radius:number
        self.radius = radius #radius store in self.radius

    def area_circle(self):
        return f"the area of circle is: {3.14 * (self.radius ** 2)}"
```

```python
c1 =  Circle(3.2)
c2 =  Circle(5.5)
c3 =  Circle(7.2)
```

```python
print(c1.area_circle())
print(c2.area_circle())
print(c3.area_circle())
```

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area_circle(self):
        return f"Area of circle is: {round(3.14 * (self.radius ** 2), 2)}"    #r

    def circumference(self):
        return f"Circumference of circle is: {round(2 * 3.14 * self.radius, 2)}"
```

```python
c1 = Circle(5)

print(c1.area_circle())
print(c1.circumference())
```

```python
c2 = Circle(5.10)

print(c2.area_circle())
print(c2.circumference())
```
