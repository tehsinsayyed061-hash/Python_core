```python
#Take number from users until user insert negative number then break 
#then loop and do sum of all previous number and do sum

while True:  # --permenant loop work karte rahega    #true: continuous
    num = int(input("Enter a number: "))
    print(num)
    if num < 0:
        print("Oop's negative number!")
    
        break
  

    

```

```python
#then loop and do sum of all previous number and do sum
sum_prev_num = 0
while True:  # --permenant loop work karte rahega
    num = int(input("Enter a number: "))
    print(num)
    if num < 0:
        print("Oop's negative number!")
        print("total sum of prev: ",sum_prev_num)
        break
    else:
        sum_prev_num += num
```

```python
import random  
secreat_number = random.randint(1,10)
while True:
    num = int(input("Guess the number 1 to 10:"))
    if num == secreat_number:
        print("Yesss~~")
        break
    elif num > secreat_number:
        print("Guess Lower:")
    elif num < secreat_number:
        print("Guess Higher:")  
    else:
         ("Sad Try again!")
    
    
```

```python
import random  
secreat_number = random.randint(1,10)
num_chance = 5
while True:
    num = int(input("Guess the number 1 to 10:"))
    num_chance -= 1

    if num == secreat_number:
        print("Yesss~~")
        break
    elif num_chance == 0:
        print("No chance left!")
        break
        
    elif num > secreat_number:
        print(f"Guess Lower: you have{ num_chance}")
    elif num < secreat_number:
        print(f"Guess Higher: you have{ num_chance}")  
    else:
         ("Sad Try again!")
```

```python
num = int(input("Enter a number: "))
fact_num = 1
while num >= 1:
    fact_num *= num
    num -= 1
print(fact_num)
    
    
```
