```python
#1 even / odd
number = int(input("Enter a number:"))

if number % 2 == 0:           # Comparision op use
    print(f"the number is{number} is even")
else:
    print(f"the number is{number} is even")
```

```python
#2 positive, negative, zero

number = int(input("Enter a number:"))

if number > 10:
    print(f"the number is {number} is Positive")
elif number == 0:                           #its also a condition case
    print(f"the number is {number} is Zero")
else:                                     # here no condition
    print(f"the number is {number} is Negative")
```

```python
#3 take 3 number from user, and fnd which one is greater

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a > b and a > c:
    print("First number is greater")
elif b > a and b > c:
    print("Second number is greater")
elif c > a and c > b:
    print("Third number is greater")
else:
    print("Two or more numbers are equal")
```

```python
#4 take a number from user, and check  the number divisible by  5 and 11 or not
num = int(input("Enter a number: "))

if num % 5 == 0 and num % 11 == 0:
    print("The number is divisible by both 5 and 11")
else:
    print("The number is not divisible by both 5 and 11")
```

```python
#5 calculate: take 2 number and 1 operator based on perform the calculation

num1 = int(input("enter firts number"))
num2 = int(input("enter second number"))
operator = input("enter any operator(+,-,*,/):")

if operator == "+":
    print(f"the sum is {num1+num2}")
elif operator == "-":
    print(f"the sum is {num1-num2}")
elif operator == "*":
    print(f"the sum is {num1*num2}")
elif operator == "/":
    print(f"the sum is {num1/num2}")
else:
    print("Invalid operator kindly use: {+,-,*,/}")
```
