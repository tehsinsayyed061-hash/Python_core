```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area_circle(self):
        return f"Area of circle is: {round(3.14 * (self.radius ** 2), 2)}"    #round:use when many decimal comes so to limit them

    def circumference(self):
        return f"Circumference of circle is: {round(2 * 3.14 * self.radius, 2)}"
```

```python
c1 = Circle(5)

print(c1.area_circle())
print(c1.circumference())
```

```python
c2 = Circle(5.10)

print(c2.area_circle())
print(c2.circumference())
```

```python
def withdraw(self, amount):
    if amount <= 0:
        print("Invalid Amount:", amount)

    elif amount > self.balance:
        print("Insufficient Balance")

    else:
        self.balance -= amount
        print("Withdrawn Amount is:", amount)
        print("Remaining Balance is:", self.balance)
```

```python
#single Inheritance
class Father:
    def father_hobby(self):
        print("Father hobby is: Reading Books")

class Child(Father):
    def child_hobby(self):
         print("Child hobby is: Playing Games")
        


```

```python
f = Father()
f.father_hobby()
```

```python
c = Child()
c.child_hobby()
c.father_hobby()  #-----method inherit from father
```

```python
#MULTIPLE INHERITANCE
class Father:
    def father_hobby(self):
        print("Father hobby is: Reading Books")
class Mother:
    def mother_hobby(self):
        print("mother hobby is: Cooking")

class Child(Father,Mother):
    def child_hobby(self):
         print("Child hobby is: Playing Games")
        
```

```python
f = Father()
f.father_hobby()
print()
m = Mother()
m.mother_hobby()
print()
c = Child()
c.child_hobby()
c.father_hobby()
c.mother_hobby()
```

```python
#MULTILEVEL INHERITANCE
class GrandFather:
    def grandfather_hobby(self):
        print("Father hobby is: Fighting")
class Father:
    def father_hobby(self,GrandFather):
        print("Father hobby is: Reading Books")

class Child:
    def child_hobby(self,Father,GrandFather):
         print("Child hobby is: Playing Games")
```

```python
g = GrandFather()
g.grandfather_hobby()
print()

f = Father()
f.grandfather_hobby()
f.father_hobby()
print()

c = Child()
c.child_hobby()
c.father_hobby()
c.grandfather_hobby()
```

```python
class Amazon:
    def __init__(self,p_name,price,offer):
        self.__p_name = p_name     #__ means private attributes but whenever we us ethis we just call them and if we made second class like inherite so it wont work
        self.__price = price
        self.__offer = offer

    def prod_details(self):
        print("Name :",self.__p_name)
        print("Price :",self.__price)
        print("Offer :",self.__offer)

    def _change_funct(self,new_price):   #_ means protected 
        self.__price = new_price
        
```

```python
re -- RegEx -- Regular Expression

\b -- border
\d -- any digit--0-9
\w-- any character a-z,A-Z,0-9,_
\s --white space to detect
. ---any char except new line
+ -- one char or more than one char
* -- 0 char or more  
? -- 0 char or one char
^ -- starting with
$ -- ending with
[0-9] -- digit between 0 to 9
[ABC] -- only char ABC
[^abc] --- any char, except a,b,c
```

```python
import re   # re : module

text = """HII am Tehsin and my Phone number is
         9984291241, and my alternate number is
         1233456789, my pin number is 123456"""

numbers = re.findall(r'\b\d{10}\b', text)
#         finall:items           r:string \b:border

print(numbers)
```
