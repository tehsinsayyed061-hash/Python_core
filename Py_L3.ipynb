```python
#Logical op:  (yeh < less than hai)
#            and,or,not

a = 5
b = 15
c = 10
print(a<b and b<c)

a = 5
b = 15
c = 10
print(a<b and b>c)

a = 5
b = 15
c = 10
print(a<b or b<c)

a = 5
b = 15
c = 10
print(not(a<b and b<c))

a = 5
b = 15
c = 10
print(not(a<b and b>c))
```

```python
#Membership op:(use in loops)
#            in, not in

a = "Tehsin"
print("i" in a)

a = "Tehsin"
print("I" in a)

a = "Tehsin"
print("i" not in a)

a = "Tehsin"
print("I" not in a)
```

```python
#Identity op:(check memory location) (value similar hai)
 #           is, is not

a = [1,2,3]
b = [1,2,3]
c = a
print(a)
print(b)
print(c)
print()

print(a==b)            # check value similar hai
print(b==c)
print(a==c)
print()

print(a is b)           #check memory location
print(b is c)
print(a is c)
print()

print(a is not b)           #check memory location
print(b is not c)
print(a is not c)
```
