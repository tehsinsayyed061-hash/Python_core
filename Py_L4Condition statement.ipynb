```python
print("-------------To Check Number is Pos, Neg, OR Zero ---------")
num = int(input("Enter a number:"))

if num > 0:
    print("Positive Number")
elif num == 0:                           #its also a condition case
    print("Number is Zero")
else:                                     # here no condition
    print("Negative Number")
```

```python
# MORE METHODS

num = int(input("Enter a number:"))

if num>0:
    print(f"the given number {num} is postive and")
    if num % 2 == 0:
        print(f"the given number{num} is Even Number")
    else:
        print(f"the given number {num} id Odd Number")
elif num ==0:
    print(f"the given number {num} is Zero")
else:
    print(f"the given number {num} is Negative")
    
```
