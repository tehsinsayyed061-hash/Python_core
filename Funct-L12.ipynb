```python
yeh keys ko call kara hai
def info(**details):
    for i in details:
        print(i)

info(name="Tehsin",age=20)
```

```python
def info(**details):
    for keys in details:
        print(keys)

info(name="Tehsin",age=20)
```

```python
value ko call karna hai tau
def info(**details):
    for values in details.values:
        print(values)

info(name="Tehsin",age=20)
```

```python
def info(**details):
    for key,values in details.items:
        print(values)

info(name="Tehsin",age=20)
```

```python
def counting(n):
    if n == 1:
        return 1
    print(n)
    return counting(n-1)
```

```python
counting(5)
```

```python
#using recursive funct we find factorial = 5*4*3*2*1

def fact(number):
    if number == 1:  # condition: base case 5=1 x 4=1 x... 1=1 yes
        return 1
    return number * fact(number -1)   #recursive case 5 * 1=5, 5-1=4; 4*1, 4-1=3;....stop cause^
fact(5)
```
